```python
def myfactorial(n):
#base case factorial of 0is 1
 if n == 1:
    return 1
#recursive case: factorial of n is multipiled by factorial of n-1
 else:
    return n*myfactorial(n-1)
```

```python
num = int(input("Enter a number to find its factorial:"))
result = myfactorial(num)
print("factorial of", num, "is", result)
```

bank customer transaction management using function
objective:
.Use a dictionary to store and retrieve customer details inbanking system
.implement function to add,update,and display customa

```python
#step 1: define a dictionary to store customeer details 
customers = {}
```

```python
#step 2: function to add a new customer
def add_customer(account_number, name, balance):
    if account_number in customer:
        print("account already exists!")
    else:
        customers[account_number] = {"name": name, "Balance": balance}
        print("customer added successfully!")        
```

```python
#step 3: adding customer
add_customer(109, "romi", 10000)
```

```python
#step 4: function to display customer details

def display_customer(account_number):
    
    if account_number in customers:
         print(f"Account Number: {account_number}")
         print(f"Name: {customers[account_number]['Name']}")
         print(f"Balance: Rs {customers[account_number]['Balance']}")
    else:
        print("Customer not found")
        
```

```python
#step 5: Displaying customer details
display_customer(109)
```

```python
add_customer(105, "pritesh", 5000)
add_customer(103, "romi", 6000)
add_customer(104, "rahul", 8000)
```

```python

#step 6: function to update customer balance

def update_balance(account_number, amount, transaction_type):
    if account_number in customers:
        if transaction_type == "deposit":
            customer[account_number]["Balance"] += amount
        elif transaction_type == "Withdraw" and customer[account_number]["Balace"] >= amount:
            customers[account_number]["Balance"] -= amount
        else:
            print("Insufficinet balance!")
        print("Transaction successfully!")
    else:
        print("Account not found!")
```

```python
#step 7 updating balance
update_balance(106, 500, "deposit")
display_customer(106)
```
