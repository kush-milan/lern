```python
months = ['jan', 'fab', 'mar', 'Apr', 'may', 'jun']
recovery_rate = [65, 70, 60, 78, 80, 85]
plt.plot(months, recovery_rate, marker ='o', linestyle='-', color='b', label="Recovery Rate")
plt.xlabel("months")
plt.ylabel("Recovery precentage")
plt.title("patient recovery Trends")
plt.legend()
plt.show()
```

```python
# BAR Plot
```

```python
medicines = ["paracetamol", "Antibiotics", "painkillers","insulin", "Antacids"]
demand = [500, 700, 450, 300, 650]
plt.bar(medicines, demand, color = ['red', 'blue', 'green', 'purple', 'orange'])
plt.xlabel("Medicine Typpe ")
plt.ylabel("unit sold")
plt.title("Medicine Demand Analysis")
plt.show()
```

```python
# pie chart
```

```python
diseases = ["Diabetes", "Hypertension", "Heart Disease", "Asthma", "others"]
percentages = [30, 25, 15, 10, 20]
colors = ["blue", "red", "green", "purple", "orange"]

plt.pie(percentages, labels = diseases, autopct ='%1.1f%%', colors = colors)
plt.title("Disease Distribution in city")
plt.show()
```

```python
# Scatter plot
```

```python
time_spent = [10, 15, 20, 25, 30, 35, 40]
diagnosis_accuracy = [60, 65, 70, 75, 80, 85, 90]

plt.scatter(time_spent, diagnosis_accuracy, color = 'red')
plt.xlabel("Time spent, by Doctor (minutes)")
plt.ylabel("Diagnosis Accuracy (%)")
plt.title("Doctor-patient interaction Time vs Diagnosis Accuracy")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Telecom churn.csv")
df.head()
```

```python
plt.figure(figsize = (7, 4))
sns.histplot(df['account length'], bins = 20, kde= True, color = 'blue')
plt.title("Distribution of call Durations")
plt.xlabel("call Duration (Minutes)")
plt.ylabel("frequency")
plt.show()
```
