```python
data = np.array([
    ["c001", 5, "Excellent service and friendly staff"],
    ["c002", 3, "Excellent service and friendly staff"],
    ["c003", 4, "Excellent service and friendly staff"],
    ["c004", 6, "Excellent service and friendly staff"],
    ["c005", 7, "Excellent service and friendly staff"],
    ["c006", 8, "Excellent service and friendly staff"],
    ["c007", 9, "Excellent service and friendly staff"],
])
```

```python
print(data)
```

```python
# step 2: extract numerical ratingsfor analysis
ratings = data[:, 1].astype(int) #convert rating column to integers
print(f"Total Feedback Entries: {len(ratings)}")
```

```python
ratings
```

```python
# step 3: perform statistical Analysis
average_rating = np.mean(ratings)
print(f"Average Customer Rating: {average_rating:.2f}")
```

```python
highest_rating = np.max(ratings)
print(f"Highest Rating Received: {highest_rating}")

lowest_rating = np.min(ratings)
print(f"Lowest Rating Received: {lowest_rating}\n")
```

```python
unique_ratings, rating_counts = np.unique(ratings, return_counts=True)
```

```python
feedback_comments = data[:, 2]

# step 5: display insights
print("\n** customer feedback analysis summary**")
print()

print("**rating distribution:**")
print()
for rating, count in zip(unique_ratings, rating_counts):
    print(f"Rating {rating}: {count} customers")
```

```python
import numpy as np
import pandas as pd
```

```python
# step 1: create a simple dataset representing insurance policyholder records

# columns: policy number, premium amount(incorrect), insured amount, claim amount (incorrect)

policy_data = np.array([
    [1001, 12000, 500000, 5000],
    [1002, 11000, 530000, 6000],
    [1003, 13000, 520000, 7000],
    [1004, 20000, 510000, 8000],
    [1007, 210000, 540000, 9000],
])

col = ["Policy Number", "Premium Amount", "Insured Amount", "Claim Amount"]
# print original dataset
print("original policyholder data:")
print()
df_before = pd.DataFrame(policy_data, columns = col)
print(df_before)
```

```python

policy_data[:, [1, 3]] = policy_data[:, [3, 1]]

print("\ncorrected policyhoder data(after wapping premium and claim amount columns):")
print()
df_after = pd.DataFrame(policy_data, columns = col)
print(df_before)
```
